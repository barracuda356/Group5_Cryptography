```python
# for part (2)
import requests
url='http://www.gutenberg.org/files/2701/2701-0.txt'
moby=requests.get(url).text
lower_moby= moby.lower()
import re
moby_new = re.sub(r'[^a-z\n]', '', lower_moby)

# for part (3)
characters= ["a" , "b", "c", "d", "e", "f", "h", "i", "j", "k", "l", "m", "n", "o", "p", "q", "r", "s", "t", "u", "v", "w", "x", "y", "z", " "]
for i in range(26):
    frequency= moby_new.count(characters[i])
    print(frequency)
```
