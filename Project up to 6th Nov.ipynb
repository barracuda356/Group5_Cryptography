```python
from string import ascii_lowercase    #this saves us having to write all the letters out

#takes in a string (could be a large example of english or a specific encoded piece of text)
#calculates the fraction of the text each letter makes up
#returns a list containing tuples of each letter and its frequency, sorted by frequency
def get_frequency(text):
    characters = ascii_lowercase+" "
    length = len(text)
    frequency = []

    for char in characters:
        frequency.append((char,text.count(char)/length))
    
    return sorted(frequency, key=lambda x: x[1])[::-1]    #kinda complicated but sorts the list by the second entry in each tuple and then reverses
    
print(get_frequency(moby_new))

#takes in a string to be decoded and a dictionary with the encoded letters as keys and their respective decoded letters as values
#returns a decoded string
def decode(ciphertext,letter_map):
    out = ""
    for char in ciphertext:
        out += letter_map[char]

    return out
    
#takes in an encoded string and list containing tuples of each letter and its frequency, sorted by frequency 
#(can be got from get_frequency)
#returns a dictionary mapping the most common letters in the encoded string with the most common letters in english
def get_letter_map_from_frequency(ciphertext,english_frequencies):
    out = {}
    text_freq = get_frequency(ciphertext)
    for i in range(len(text_freq)):
        encoded_letter = text_freq[i][0]
        english_letter = english_frequencies[i][0]

        out[encoded_letter] = english_letter

    return out
```
